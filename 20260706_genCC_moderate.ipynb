```python
import pandas as pd

gencc = pd.read_csv("gencc-submissions.csv", low_memory=False)

# Filter to Definitive, Strong, or Moderate classifications
keep = ["Definitive", "Strong", "Moderate"]
gencc_filtered = gencc[gencc["classification_title"].isin(keep)]

# Deduplicate to one row per gene symbol
unique_gencc = gencc_filtered[["gene_symbol"]].drop_duplicates().reset_index(drop=True)

unique_gencc.to_csv("unique_gencc-submissions_20260706.csv", index=False)

print(f"Rows in filtered data: {len(gencc_filtered)}")
print(f"Unique genes: {len(unique_gencc)}")
unique_gencc.head()
```

```python
old_gencc = pd.read_csv("unique_gencc.csv", index_col=0)

old_genes = set(old_gencc["gene_symbol"].dropna())
new_genes = set(unique_gencc["gene_symbol"].dropna())

in_both        = new_genes & old_genes
in_new_only    = new_genes - old_genes
in_old_only    = old_genes - new_genes

print(f"Old unique_gencc.csv:              {len(old_genes):>5} genes")
print(f"New unique_gencc-submissions:      {len(new_genes):>5} genes")
print(f"")
print(f"In both:                           {len(in_both):>5} genes")
print(f"In new only (added vs old):        {len(in_new_only):>5} genes")
print(f"In old only (dropped from new):    {len(in_old_only):>5} genes")
```

```python
dupes = unique_gencc[unique_gencc.duplicated(subset="gene_symbol", keep=False)]
print(f"Duplicate rows: {len(dupes)}")
if len(dupes) > 0:
    print(dupes)
```
