```python
import pandas as pd
```

```python

maaveable_df = pd.read_csv('MAVEable_final_is_feasible.csv')
maaveable_df
```

```python
# Manual gene list to check against the MAVEable dataset
manual_genes = [
    "APP", "PSEN1", "PSEN2", "SORL1", "MAPT", "ADAM10",
    "LRRK2", "GBA1", "TMEM175", "VPS13C", "ATP13A2", "RAB32",
    "SNCA", "PRKN", "PINK1", "PARK7", "VPS35", "SYNJ1", "DNAJC6",
    "CHCHD2", "SH3GL2", "RIT2", "SV2C",
    "TARDBP", "C9orf72", "FUS", "SOD1", "GRN", "SQSTM1", "OPTN",
    "VCP", "UBQLN2", "CHCHD10", "ANXA11", "TBK1", "UNC13A", "ATXN2",
    "KIF5A", "PFN1", "TUBA4A", "NEFH", "DCTN1", "SETX", "MATR3",
    "HNRNPA1", "HNRNPA2B1", "ITSN1"
]
```

```python
# Check which genes are in the original MAVEable dataset
maave_genes = set(maaveable_df['GeneSymbol'])
in_maveable = [g for g in manual_genes if g in maave_genes]
not_in_maveable = [g for g in manual_genes if g not in maave_genes]

print(f"Manual genes: {len(manual_genes)}")
print(f"In MAVEable dataset: {len(in_maveable)}\n{in_maveable}")
print(f"\nNOT in MAVEable dataset: {len(not_in_maveable)}\n{not_in_maveable}")
```

```python
# Filter: remove manual gene list, apply length < 592, then add them back
maave_filtered = maaveable_df[~maaveable_df['GeneSymbol'].isin(in_maveable)].copy()
maave_filtered_len_lt592 = maave_filtered[maave_filtered['Length'] < 592]

# Add back the 39 manual genes
maave_final = pd.concat([maave_filtered_len_lt592, maaveable_df[maaveable_df['GeneSymbol'].isin(in_maveable)]])

print(f"Original MAVEable: {len(maaveable_df)}")
print(f"After removing {len(in_maveable)} manual genes: {len(maave_filtered)}")
print(f"After length < 592 filter: {len(maave_filtered_len_lt592)}")
print(f"Final (with {len(in_maveable)} genes added back): {len(maave_final)}")
```

```python

# Bring in the 7-gene file and append to maave_final
seven_df = pd.read_csv('MAVEable_seven_genes_20260917.csv')

# Check columns match
missing_cols = set(maave_final.columns) - set(seven_df.columns)
extra_cols = set(seven_df.columns) - set(maave_final.columns)

if missing_cols:
    print(f"Missing in 7-gene file: {missing_cols}")
if extra_cols:
    print(f"Extra in 7-gene file: {extra_cols}")

# Ensure same columns (add any missing as NaN)
for col in maave_final.columns:
    if col not in seven_df.columns:
        seven_df[col] = None

# Skip seven-gene rows for genes already in maave_final (e.g. RAB32, UNC13A now pass the
# main pipeline filters with the 2026-09-11 GenCC); keep the filtered main-pipeline counts
already_present = seven_df['GeneSymbol'].isin(maave_final['GeneSymbol'])
print(f"Already in maave_final, not appended again: {seven_df.loc[already_present, 'GeneSymbol'].tolist()}")
seven_df = seven_df[~already_present]

maave_final_with_7 = pd.concat([maave_final, seven_df[maave_final.columns]]).reset_index(drop=True)
print(f"maave_final rows: {len(maave_final)}")
print(f"seven_genes rows: {len(seven_df)}")
print(f"Combined total: {len(maave_final_with_7)}")
```

```python

# Write the final combined dataframe to CSV
maave_final_with_7.to_csv('Broad_scan_list_20260917.csv', index=False)
print(f"Written {len(maave_final_with_7)} rows to Broad_scan_list.csv")

```
